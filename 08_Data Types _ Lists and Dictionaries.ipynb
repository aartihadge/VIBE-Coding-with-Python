```python
nums = [1, 4, 15, 456]  # Example list of numbers
# Determine maximum even
max_num = None
for num in nums:
    if (max_num == None) and (num % 2 == 0):
        # First even number found
        max_num = num
    elif (max_num != None) and (num > max_num) and (num % 2 == 0):
        # Larger even number found
        max_num = num
print(f'Max even number: {max_num}')
```

##### Iterating over multi-dimensional lists


```python
currency = [

       [1.00, 5.00, 10.0], # US Dollars 
       [0.65, 3.25, 6.50], # British pounds
       [0.75, 3.77, 7.53], # Euros

]
```

```python
for row in currency:
    print('row', row)  # Print each row of the two-dimensional list
    for col in row:
        print(col, end=' ')
    print()  # New line after each row
```

##### Iterating through multi-dimensional lists using enumerate()

```python
for row_index, row in enumerate(currency):
    print(f'Row {row_index}:', row, end=' ')  # Print each row of the two-dimensional list
    for col_index, col in enumerate(row):
        print(f'Column {col_index}: {col}', end=' ')
    print()  # New line after each row
```

```python
# Get a list of integers from the user
numbers = [int(i) for i in input('Enter numbers:').split()]
print('Original list:', numbers)

# Return a list of only even numbers
even_numbers = [i for i in numbers if (i % 2)==0]
print('Even numbers only:', even_numbers)
```

##### Using sorted() to create a new sorted list from an existing list without modifying the existing list

```python
print('Sorted number:', sorted_numbers)
```

#### Using the key argument

```python
names = ['Ram', 'shyam', 'Krishna', 'laxman']
print(names)
```

```python
no_key_sort = sorted (names)
print('Sorting without key:',no_key_sort)
```

```python
key_sort = sorted (names, key=str.lower)
print('Sorting with key:', key_sort)
```
