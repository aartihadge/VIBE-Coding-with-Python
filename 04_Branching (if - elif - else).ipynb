### Implict Range


```python
x = int(input('Enter the value Of x: ' ) )
print(f'You entered {x}.')

if x < 0 :
   print( 'Negative' )
elif (x <=10) :      # x>=0 is implicit
   print('0..10')
elif (x <= 20) :     # x > 10 is implicit
   print('11..20')
else :               # x>20 is implicit
   print('21')
```
